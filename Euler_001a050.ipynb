#Problemas de 11 a 20

###Largest Product in a Grid

####Problem 11
In the 20 x 20 grid (dataset provided in inputs/matriz_p11.txt), four numbers along a diagonal line were multiplied (starting at row 7, column 9).

The product of these numbers is 26 x 63 x 78 x 14 = 1788696.

What is the greatest product of four adjacent numbers in the same direction (up, down, left, right, or diagonally) in the 20 x 20 grid?

```python
from lib_11a20 import prod_direcao

arq = open("inputs/matriz_p11.txt", "r")
matriz = arq.readlines()
prod = 1

for i, linha in enumerate(matriz):
  matriz[i] = linha.replace("\n", "").split(" ")
  for j, elemento in enumerate(matriz[i]):
    matriz[i][j] = int(elemento)

for i, linha in enumerate(matriz):
  for j in range(len(linha)):
    produto = prod_direcao(matriz, i, j, 4)
    prod = max(prod, produto)

print(prod)
```

###Highly Divisible Triangular Number

####Problem 12
The sequence of triangle numbers is generated by adding the natural numbers. So the 7th triangle number would be 1 + 2 + 3 + 4 + 5 + 6 + 7 = 28. The first ten terms would be:

1, 3, 6, 10, 15, 21, 28, 36, 45, 55, ...

Let us list the factors of the first seven triangle numbers:

1: 1

3: 1, 3

6: 1, 2, 3, 6

10: 1, 2, 5, 10

15: 1, 3, 5, 15

21: 1, 3, 7, 21

28: 1, 2, 4, 7, 14, 28

We can see that 28 is the first triangle number to have over five divisors.

What is the value of the first triangle number to have over five hundred divisors?

```python
from lib_11a20 import primos, num_divisores, triangular

lista_primos = primos(100000)
n = 1
result = 1

while True:
  result = triangular(n)
  if (num_divisores(result, lista_primos) > 500):
    break
  n += 1

print(result)
```

###Longest Collatz Sequence

####Problem 14
The following iterative sequence is defined for the set of positive integers:

n --> n/2  (n is even)

n --> 3n + 1  (n is odd)

Using the rule above and starting with 13, we generate the following sequence:

13 --> 40 --> 20 --> 10 --> 5 --> 16 --> 8 --> 4 --> 2 --> 1.

It can be seen that this sequence (starting at 13 and finishing at 1) contains 10 terms. Although it has not been proved yet (Collatz Problem), it is thought that all starting numbers finish at 1.

Which starting number, under one million, produces the longest chain?

NOTE: Once the chain starts the terms are allowed to go above one million.

```python
from lib_11a20 import collatz
result = collatz(1000000)
print(result.index(max(result)))
```
